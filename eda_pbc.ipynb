```python
import pandas as pd
import numpy as np
import seaborn as sns

filename = 'pbc.csv'

pbc = pd.read_csv(filename)
```

```python
pbc.head()
```

```python
pbc.hist()
```

```python
variables_to_log_transform = ['bili', 'chol', 'alk.phos', 'ast', 'trig', 'protime']

# these were chosen, because they're the ones who appear to have a long rhs skew to the data
# the resulting histograms tend to move towards a more gaussian distribution, which i think is good
```

```python
for variable in variables_to_log_transform:
    new_variable = 'log_' + variable
    pbc[new_variable] = np.log(pbc[variable])

    pbc.hist(column = [variable, new_variable])
```
